```python
import logging
from logging.handlers import RotatingFileHandler
import os
import datetime

# Create log directory if needed
os.makedirs("logs", exist_ok=True)

# Configure RotatingFileHandler
log_file = "logs/fastAPI.production-model.csv"
handler = RotatingFileHandler(
    log_file,
    maxBytes=1024 * 10,  # 10 KB
    backupCount=10       # Keep 10 rotated logs
)

# Define a formatter with a custom date format
formatter = logging.Formatter(
    fmt='%(asctime)s,%(levelname)s,%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)
handler.setFormatter(formatter)

# Create the logger
logger = logging.getLogger("fastAPI.production-model")
logger.setLevel(logging.DEBUG)  # Set minimum log level
logger.addHandler(handler)

# Optional: avoid duplicate logs if root logger has handlers
logger.propagate = False

# Define your CSV-style message builder
def makeMessage(currentTimeUnix, errorType, sourceService, apiEndpoint, val1, val2, message, request_ID=''):
    return f"{currentTimeUnix},{errorType},{sourceService},{apiEndpoint},{val1},{val2},{message},{request_ID}"

# Example log usage
current_time_unix = int(datetime.datetime.utcnow().timestamp() * 1000)
log_msg = makeMessage(
    currentTimeUnix=current_time_unix,
    errorType="OK",
    sourceService="fastAPI.production-model",
    apiEndpoint="",
    val1="1.26",
    val2="None",
    message="FastAPI production model running normally at 1.26%",
    request_ID="4a2ca74d-778c-484f-ba38-9e1bd546420c"
)

logger.info(log_msg)
```

```python
import picologging
from logging.handlers import RotatingFileHandler
import csv
import os
import datetime

# Create log directory if needed
os.makedirs("logs", exist_ok=True)

# Configure RotatingFileHandler
log_file = "logs/my_log.csv"
handler = RotatingFileHandler(
    log_file,
    maxBytes=1024 * 10,  # 10 KB for example
    backupCount=10        # Keep 3 rotated logs
)

# 2025-05-31 15:51:37,INFO,1748721097979,OK,fastAPI.production-model,,1.26,None,FastAPI production model running normally at 1.26%,4a2ca74d-778c-484f-ba38-9e1bd546420c

def makeMessage(currentTimeUnix, errorType, sourceService, apiEndpoint, val1, val2, message, request_ID=''):
    return f"{currentTimeUnix},{errorType},{sourceService},{apiEndpoint},{val1},{val2},{message},{request_ID}"

# Define formatter with custom date format (e.g. YYYY-MM-DD HH:MM:SS)
formatter = picologging.Formatter(
    fmt='%(asctime)s,%(levelname)s,%(message)s',    
    datefmt='%Y-%m-%d %H:%M:%S'
)

handler.setFormatter(formatter)

# Create logger
logger = picologging.getLogger("my_csv_logger")
logger.setLevel(picologging.DEBUG)  # You can change to INFO, WARNING, etc.
logger.addHandler(handler)

logMessage = makeMessage(
    currentTimeUnix=int(datetime.datetime.now().timestamp() * 1000),  # Current time in milliseconds
    errorType="OK",
    sourceService="fastAPI.production-model",
    apiEndpoint="",
    val1=1.26,
    val2=None,
    message="FastAPI production model running normally at 1.26%",
    request_ID="4a2ca74d-778c-484f-ba38-9e1bd546420c"
)

# Example log messages
logger.debug("Debug message")
logger.info(logMessage)
logger.warning("Warning message")
logger.error("Error message")
logger.critical("Critical message")
```
